# numerical.run on the RunMat kernel

This is a numerical.run `.nb.md` notebook running inside **Jupyter** — MATLAB
syntax on the RunMat kernel, entirely in the browser (JupyterLite) or on a
RunMat kernelspec (JupyterLab).

A quick computation and a plot:

```matlab
x = linspace(0, 2*pi, 400);
y = sin(3*x) .* exp(-0.3*x);
figure; plot(x, y, 'LineWidth', 2); title('damped sine'); xlabel('x');
```

Read a couple of values back (no semicolon → the kernel echoes them):

```matlab
peak = max(y)
energy = sum(y.^2)
```
